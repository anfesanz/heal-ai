# 03 Evaluation by Subgroup

Calculate subgroup metrics, performance gaps, and bootstrap intervals.

```python
from src.data_loader import load_or_build_nhanes_dataset
from src.modelling import train_models
from src.evaluation import evaluate_all_subgroups, evaluate_models
from src.fairness_metrics import subgroup_gap_summary
from src.preprocessing import GROUP_COLUMNS
from src.uncertainty import bootstrap_by_subgroup

df = load_or_build_nhanes_dataset(download=False)
models, X_train, X_test, y_train, y_test = train_models(df)
available_groups = [col for col in GROUP_COLUMNS if col in X_test]
subgroup_metrics = evaluate_all_subgroups(models, X_test, y_test, available_groups)
subgroup_metrics.head()
```

```python
gaps = subgroup_gap_summary(subgroup_metrics)
gaps.sort_values('gap', ascending=False).head(10)
```

```python
overall = evaluate_models(models, X_test, y_test)
best_model_name = overall.iloc[0]['model']
best_prob = models[best_model_name].predict_proba(X_test)[:, 1]
bootstrap_by_subgroup(y_test, best_prob, X_test[available_groups[0]], best_model_name, available_groups[0], n_bootstrap=100).head()
```
